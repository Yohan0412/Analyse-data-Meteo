```python
#Voir si la distribution des données est normal 
from scipy.stats import shapiro , probplot

# Test shapiro pour vérifier la normalité de la distribution
stat, p = shapiro(df['conso_gaz_mw'])
print("Statistique", stat)
print("p-value", p)

if p > 0.05:
    print("La distribution est normale (p > 0.05)") 
else:
    print("La distribution n'est pas normale (p <= 0.05)")
    
    #---------------------------------------------------------------
#Q-Q Plot

plt.figure(figsize=(6, 6))
probplot(df['conso_gaz_mw'], dist="norm", plot=plt)
plt.title("Q-Q Plot de Performance")
plt.grid(True)
plt.tight_layout()
plt.show()
```

## La distribution des données est pas normal on va donc utilisé spearman au lieux de Pearson pour calculer la corrélation.

```python
# Corrélation entre Température et Consommation
from scipy.stats import spearmanr

corr, p_value = spearmanr(df['temperature_2m'], df['conso_gaz_mw'])

print("Coefficient de Spearman :", corr)
print("P-value :", p_value)
```

# Corrélation entre Température et Consomation Gaz :

La corrélation est de -0. 55 ce qui signifie que plus le température baisse plus la consomation de Gaz augmente 
, probablement en raison de l’utilisation du chauffage.

```python
from sklearn.linear_model import LinearRegression
import numpy as np

X = df[['temperature_2m']].values  # Variable explicative (2D)
y = df['conso_gaz_mw'].values        # Variable à expliquer

model = LinearRegression()
model.fit(X, y)

print("Coefficient (impact température) :", model.coef_[0])
print("Intercept :", model.intercept_)
print("Score R² :", model.score(X, y))
```

## Régression linéaire simple :

Coéfficient : -261 ce qui signifie que pour chaque augmentation de 1°C la consomation baisse en moyenne de 261 Mw


Intercept : La valeur estimée de la consomation quand la température est de 0°C est de 7606 Kw

Le Score R^2 indique de 27% de la variance de la consomation est expliquée par la température

```python
import matplotlib.pyplot as plt
import pandas as pd

# S'assurer que datetime est bien au format datetime
df["datetime"] = pd.to_datetime(df["datetime"])
df["month"] = df["datetime"].dt.month


conso_gaz_moyenne = df.groupby("month")["conso_gaz_mw"].mean()
temp_moyenne = df.groupby("month")["temperature_2m"].mean()


fig, ax1 = plt.subplots(figsize=(10, 5))


ax1.bar(conso_gaz_moyenne.index, conso_gaz_moyenne, color="lightcoral", label="Conso Gaz (MW)")
ax1.set_xlabel("Mois")
ax1.set_ylabel("Consommation de gaz (MW)", color="lightcoral")
ax1.tick_params(axis='y', labelcolor="lightcoral")
ax1.set_xticks(range(1, 13))
ax1.set_xticklabels(["Jan", "Fév", "Mar", "Avr", "Mai", "Juin", "Juil", "Août", "Sept", "Oct", "Nov", "Déc"])

# Courbe pour la température 
ax2 = ax1.twinx()
ax2.plot(temp_moyenne.index, temp_moyenne, color="steelblue", marker="o", label="Température (°C)")
ax2.set_ylabel("Température moyenne (°C)", color="steelblue")
ax2.tick_params(axis='y', labelcolor="steelblue")

# Titre et mise en page
plt.title("Consommation de gaz et température moyenne par mois")
fig.tight_layout()
plt.show()
```

## Consomation de Gaz par mois

On voit sur le graphique que les **fortes consommations de gaz** se situent durant les mois les plus froids, notamment en **Janvier, Février, Novembre et Décembre.**

À l’inverse, la consommation **diminue fortement** pendant les mois les plus chauds, c’est-à-dire en **Juin, Juillet et Août.**

On observe également que plus la courbe des **températures est basse**, plus la **consommation augmente**.
